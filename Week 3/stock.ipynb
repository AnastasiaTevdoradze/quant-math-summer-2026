Day 1- Build a stock class. Attributes- ticker, prices, shares. 
Method- market_value(): returns price x shares 

Day2- Add two subclasses for ETF and Option with expense ratio and in the money methods. 

```python
class Stock:

    transaction_fee = 0.001 #class variable 

    def __init__(self, ticker, price, shares):
        self.ticker = ticker 
        self.price = price 
        self.shares = shares 
         
    
    def market_value(self):
        return self.price * self.shares

    def trade_cost(self):
        return self.shares * self.price * (1 + self.transaction_fee)



class ETF(Stock):
    def __init__(self, ticker, price, shares, holdings=None):
        super().__init__(ticker, price, shares)
        self.holdings = holdings 
    

    def expense_ratio(self): 
        return 0.0003
    


class Option(Stock): 
    def __init__(self, ticker, price, shares, strike_price): 
        super().__init__(ticker, price, shares)
        self.strike_price = strike_price


    def in_the_money(self):
            return self.price > self.strike_price

aapl_option = Option("AAPL", 182.00, 1, 175.00)
print(aapl_option.in_the_money())   # should print True

```
