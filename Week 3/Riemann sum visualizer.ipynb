```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 3, 500)

def f(x):
    return x**2

def riemann_sum(n):
    a = 0 
    b = 3
    width = (b - a) / n 
def riemann_sum(n):
    a = 0
    b = 3
    width = (b - a) / n
    x_left = np.linspace(a, b, n, endpoint=False)
    height = f(x_left)
    return width * np.sum(height)

print(f"n=4: {riemann_sum(4): .4f}")
print(f"n=10: {riemann_sum(10): .4f}")
print(f"n=50: {riemann_sum(50): .4f}")
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

ns = [4, 10, 50]

for i, n in enumerate(ns): 
    ax = axes[i]
    ax.plot(x, f(x), color='steelblue', linewidth=2)
    a, b = 0, 3
    width = (b - a) / n
    x_left = np.linspace(a, b, n, endpoint=False)
    height = f(x_left)
    ax.bar(x_left, height, width=width, align='edge',
           color='steelblue', alpha=0.3, edgecolor='black', linewidth=0.5)
    ax.set_title(f"n = {n} rectangles\narea = {riemann_sum(n): .4f}", fontsize=11)
    ax.set_xlabel("x")
    ax.set_ylabel("f(x)=x^2")
    ax.grid(True, alpha=0.3)

plt.suptitle("Riemann Sum Convergning to Exact Area", fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()
```

Week 3- Riemann Sum Visualizer 

The definite integral of x^2 from 0 to 3 equals exactly 9.0 by FTC. 

The Visualizer shpws how Riemann sums approximate that area by dividing it into n rectangles. As n increases the approximation converges to the exact value - the same limiting process that defines integral.
